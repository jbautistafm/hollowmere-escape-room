```python
# define rooms and items

# OBJECTS IN GAME ROOM

mirror = {
    "name": "mirror",
    "type": "furniture",
    "hint": "Your reflection smiles. You didn't. A message appears: 'Look where the dead left their clothes.'"
}

closet = {
    "name": "closet",
    "type": "furniture",
    "hint": "You open the closet and a snake suddenly appears! Behind it, you find a note: 'Black and white, side by side...'"
}

keyboard = {
    "name": "keyboard",
    "type": "furniture",
    "hint": "You try the keys. One of them doesn't make a sound."
}

door_a = {
    "name": "door_a",
    "type": "door"
}

key_a = {
    "name": "key for door_a",
    "type": "key",
    "target": door_a
}

# OBJECTS IN BEDROOM 1

photograph = {
    "name": "photograph",
    "type": "furniture",
    "hint": "There is a family. On the back, someone wrote: 'She hated darkness.'"
}

lamp = {
    "name": "lamp",
    "type": "furniture",
    "hint": "The light flickers. For a moment, the shadow of a woman appears. Engraved on the lamp are the words: 'Where she slept, she hid everything.'"
}

queen_bed = {
    "name": "queen_bed",
    "type": "furniture",
    "hint": "You look under the mattress and notice something hidden there."
}

door_b = {
    "name": "door_b",
    "type": "door"
}

key_b = {
    "name": "key for door_b",
    "type": "key",
    "target": door_b
}

# OBJECTS IN BEDROOM 2

laptop = {
    "name": "laptop",
    "type": "furniture",
    "hint": "The screen turns on by itself. You see the same family photograph. A message appears: 'They tried to erase me. But they couldn't erase what I wrote.'"
}

bookshelf = {
    "name": "bookshelf",
    "type": "furniture",
    "hint": "You find a dusty book. Some pages were written by Isolde: 'They call me a witch... If you're reading this, you're trapped here too. I hid what you need where I spent my last night.'"
}

bed = {
    "name": "bed",
    "type": "furniture",
    "hint": "Under the pillow, you find a note: 'The truth is waiting downstairs.'"
}

door_c = {
    "name": "door_c",
    "type": "door"
}

key_c = {
    "name": "key for door_c",
    "type": "key",
    "target": door_c
}


# OBJECTS IN LIVING ROOM

clock = {
    "name": "clock",
    "type": "furniture",
    "hint": "The clock has stopped. Its hands point to 3:17. Engraved underneath are the words: 'Remember my time.'"
}

closet_living_room = {
    "name": "closet_living_room",
    "type": "furniture",
    "code": "317"
}

door_d = {
    "name": "door_d",
    "type": "door"
}

key_d = {
    "name": "key for door_d",
    "type": "key",
    "target": door_d
}


# ROOMS

game_room = {
    "name": "game_room",
    "type": "room"
}

bedroom_1 = {
    "name": "bedroom_1",
    "type": "room"
}

bedroom_2 = {
    "name": "bedroom_2",
    "type": "room"
}

living_room = {
    "name": "living_room",
    "type": "room"
}

outside = {
    "name": "outside"
}


# LIST OF ALL ROOMS

all_rooms = [
    game_room,
    bedroom_1,
    bedroom_2,
    living_room,
    outside
]


# LIST OF ALL DOORS

all_doors = [
    door_a,
    door_b,
    door_c,
    door_d
]


# DEFINE WHICH ITEMS / ROOMS ARE RELATED

object_relations = {

    # ---------- GAME ROOM ----------
    "game_room": [
        mirror,
        closet,
        keyboard,
        door_a
    ],

    # The key for door_a is hidden in the keyboard
    "keyboard": [key_a],


    # ---------- BEDROOM 1 ----------
    "bedroom_1": [
        door_a,
        photograph,
        lamp,
        queen_bed,
        door_b
    ],

    # The key for door_b is hidden in the queen bed
    "queen_bed": [key_b],


    # ---------- BEDROOM 2 ----------
    "bedroom_2": [
        door_b,
        laptop,
        bookshelf,
        bed,
        door_c
    ],

    # The key for door_c is hidden in the bed
    "bed": [key_c],


# ---------- LIVING ROOM ----------
"living_room": [
    door_c,
    clock,
    closet_living_room,
    door_d
    ],


    # The key for door_d is hidden in the living room closet
    "closet_living_room": [key_d],


    # ---------- OUTSIDE ----------
    "outside": [door_d],


    # ---------- DOOR CONNECTIONS ----------

    # Game Room ↔ Bedroom 1
    "door_a": [
        game_room,
        bedroom_1
    ],

    # Bedroom 1 ↔ Bedroom 2
    "door_b": [
        bedroom_1,
        bedroom_2
    ],

    # Bedroom 2 ↔ Living Room
    "door_c": [
        bedroom_2,
        living_room
    ],

    # Living Room ↔ Outside
    "door_d": [
        living_room,
        outside
    ]
}

# define game state. Do not directly change this dict.
# Instead, when a new game starts, make a copy of this
# dict and use the copy to store gameplay state. This
# way you can replay the game multiple times.

INIT_GAME_STATE = {
    "current_room": game_room,
    "keys_collected": [],
    "target_room": outside
}
```

```python
# FUNCIÓN AUXILIAR: SALTO DE LÍNEA
# Esta función no recibe parámetros.
# Su única función es imprimir dos saltos de línea para separar el texto.

def linebreak():
    """
    Print a line break
    """
    print("\n\n")


# FUNCIÓN PARA EMPEZAR EL JUEGO
# Tampoco recibe parámetros.
# Primero muestra la introducción y después llama a play_room()
# usando como habitación la habitación actual guardada en game_state.

def start_game():
    """
    Start the game
    """
    print("You wake up on a couch and find yourself in a strange house with no windows which you have never been to before. You don't remember why you are here and what had happened before. You feel some unknown danger is approaching and you must get out of the house, NOW!")

    # Accedemos al valor "current_room" del diccionario game_state
    # y se lo pasamos como argumento a play_room()
    play_room(game_state["current_room"])


#PISTA

def hint(item):
    print(item["hint"])

# FUNCIÓN PRINCIPAL PARA JUGAR UNA HABITACIÓN
# Recibe una habitación como parámetro.
# Comprueba si hemos llegado al objetivo.
# Si no, permite elegir entre explorar o examinar.

def play_room(room):
    """
    Play a room. First check if the room being played is the target room.
    If it is, the game will end with success. Otherwise, let player either
    explore (list all items in this room) or examine an item found here.
    """

    # Actualizamos la habitación actual dentro del estado del juego
    game_state["current_room"] = room

    # Comprobamos si la habitación actual es la habitación objetivo
    if(game_state["current_room"] == game_state["target_room"]):

        # Si lo es, el jugador ha escapado
        print("For a second, Isolde is standing behind you.")
        print("This time, when your reflection smiles...")
        print("...you smile too.")

        print("\nCONGRATULATIONS!")
        print("YOU ESCAPED THE HOLLOWMERE HOUSE.")

    else:
        # Si todavía no hemos llegado al objetivo,
        # mostramos el nombre de la habitación actual
        print("You are now in " + room["name"])

        # Pedimos al jugador qué quiere hacer.
        # .strip() elimina espacios al principio y al final.
        intended_action = input(
            "What would you like to do? Type 'explore' or 'examine'?"
        ).strip()

        # Si escribe "explore", exploramos la habitación
        if intended_action == "explore":
            explore_room(room)

            # Después volvemos a play_room() para seguir jugando
            play_room(room)

        # Si escribe "examine", preguntamos qué objeto quiere examinar
        elif intended_action == "examine":
            examine_item(
                input("What would you like to examine?").strip()
            )

        # Si escribe otra cosa, mostramos un mensaje de error
        else:
            print("Not sure what you mean. Type 'explore' or 'examine'.")

            # Volvemos a preguntar llamando otra vez a play_room()
            play_room(room)

        # Imprimimos saltos de línea para separar visualmente el texto
        linebreak()


# FUNCIÓN PARA EXPLORAR UNA HABITACIÓN
# Recibe la habitación actual y muestra los objetos que contiene.

def explore_room(room):
    """
    Explore a room. List all items belonging to this room.
    """

    # LIST COMPREHENSION
    # 1. room["name"] obtiene el nombre de la habitación.
    # 2. object_relations[...] obtiene los objetos relacionados con ella.
    # 3. Recorremos esos objetos con "for i in..."
    # 4. De cada objeto guardamos únicamente su nombre i["name"].
    #
    # Resultado ejemplo:
    # ["couch", "piano", "door a", "closet"]

    items = [
        i["name"]
        for i in object_relations[room["name"]]
    ]

    # ", ".join(items) une los nombres de la lista usando comas.
    # Ejemplo:
    # ["couch", "piano"] -> "couch, piano"

    print(
        "You explore the room. This is "
        + room["name"]
        + ". You find "
        + ", ".join(items)
    )


# FUNCIÓN PARA SABER QUÉ HABITACIÓN HAY AL OTRO LADO DE UNA PUERTA
# Recibe:
# - door = puerta examinada
# - current_room = habitación en la que estamos
#
# Devuelve la otra habitación conectada con esa puerta.

def get_next_room_of_door(door, current_room):
    """
    From object_relations, find the two rooms connected to the given door.
    Return the room that is not the current_room.
    """

    # Buscamos en object_relations las habitaciones conectadas
    # con esta puerta.
    connected_rooms = object_relations[door["name"]]

    # Recorremos las habitaciones conectadas
    for room in connected_rooms:

        # Si esta habitación NO es la habitación actual...
        if(not current_room == room):

            # ...la devolvemos como siguiente habitación.
            return room


# FUNCIÓN PARA EXAMINAR UN OBJETO
# Esta es la función más compleja de esta celda.
#
# Recibe el NOMBRE del objeto que el jugador quiere examinar.
# Después comprueba:
# - si existe en la habitación
# - si es una puerta
# - si tenemos su llave
# - o si el objeto contiene algo

def examine_item(item_name):
    """
    Examine an item which can be a door or furniture.
    First make sure the intended item belongs to the current room.
    Then check if the item is a door. Tell player if key hasn't been
    collected yet. Otherwise ask player if they want to go to the next
    room. If the item is not a door, then check if it contains keys.
    Collect the key if found and update the game state. At the end,
    play either the current or the next room depending on the game state
    to keep playing.
    """

    # Guardamos la habitación actual en una variable más corta
    current_room = game_state["current_room"]

    # Inicialmente todavía no sabemos cuál será la siguiente habitación
    next_room = ""

    # Inicialmente todavía no tenemos ningún mensaje de resultado
    output = None


    # Recorremos todos los objetos relacionados con la habitación actual
    for item in object_relations[current_room["name"]]:

        # Comprobamos si el nombre del objeto coincide
        # con el que ha escrito el jugador
        if(item["name"] == item_name):

            # Empezamos a construir el mensaje que verá el jugador
            output = "You examine " + item_name + ". "


            # ---------- SI EL OBJETO ES UNA PUERTA ----------

            if(item["type"] == "door"):

                # Al principio suponemos que NO tenemos la llave
                have_key = False

                # Recorremos todas las llaves recogidas
                for key in game_state["keys_collected"]:

                    # Comprobamos si esta llave abre esta puerta
                    if(key["target"] == item):

                        # Si coincide, tenemos la llave correcta
                        have_key = True


                # Si tenemos la llave correcta...
                if(have_key):

                    output += "You unlock it with a key you have."

                    # Buscamos la habitación situada al otro lado
                    next_room = get_next_room_of_door(
                        item,
                        current_room
                    )

                # Si NO tenemos la llave...
                else:
                    output += "It is locked but you don't have the key."


            # ---------- SI EL OBJETO NO ES UNA PUERTA ----------

            else:

                # ---------- SI EL OBJETO NECESITA UN CÓDIGO ----------

                if "code" in item:

                    # Pedimos al jugador que introduzca el código
                    entered_code = input(
                        "Enter the code: "
                    ).strip()

                    # Comprobamos si el código es correcto
                    if entered_code == item["code"]:

                        output += "The code is correct. The closet opens. "

                        # Comprobamos si hay algo dentro
                        if (
                            item["name"] in object_relations
                            and len(object_relations[item["name"]]) > 0
                        ):

                            # Extraemos el objeto escondido
                            item_found = object_relations[item["name"]].pop()

                            # Lo añadimos a las llaves recogidas
                            game_state["keys_collected"].append(item_found)

                            # Indicamos qué hemos encontrado
                            output += "You find " + item_found["name"] + "."

                    # Si el código es incorrecto
                    else:
                        output += "Wrong code. The closet remains locked."


                # ---------- SI EL OBJETO CONTIENE ALGO ----------

                elif (
                    item["name"] in object_relations
                    and len(object_relations[item["name"]]) > 0
                ):

                    # Si tiene una pista, la mostramos primero
                    if "hint" in item:
                        output += item["hint"] + " "

                    # Extraemos el objeto escondido
                    item_found = object_relations[item["name"]].pop()

                    # Lo añadimos a las llaves recogidas
                    game_state["keys_collected"].append(item_found)

                    # Añadimos al mensaje qué hemos encontrado
                    output += "You find " + item_found["name"] + "."


                # ---------- SI EL OBJETO SOLO TIENE UNA PISTA ----------

                elif "hint" in item:

                    output += item["hint"]


                # ---------- SI EL OBJETO NO TIENE NADA ----------

                else:

                    output += "There isn't anything interesting about it."


            # Mostramos el resultado de examinar el objeto
            print(output)

            # Ya hemos encontrado el objeto, terminamos el for
            break


    # Si output sigue siendo None, no encontramos el objeto
    if(output is None):

        print(
            "The item you requested is not found in the current room."
        )


    # Si hemos abierto una puerta, preguntamos si queremos cruzarla
    if(
        next_room
        and input(
            "Do you want to go to the next room? Enter 'yes' or 'no'"
        ).strip() == 'yes'
    ):

        # Entramos en la siguiente habitación
        play_room(next_room)

    else:

        # Seguimos en la habitación actual
        play_room(current_room)
```

```python
# CREATE THE GAME STATE
# Creamos una copia de INIT_GAME_STATE para usarla durante esta partida.
# .copy() permite modificar game_state sin modificar directamente
# el estado inicial original (INIT_GAME_STATE).

game_state = INIT_GAME_STATE.copy()


# START THE GAME
# Llamamos a la función start_game() para comenzar la partida.
# No necesita argumentos porque obtiene la habitación actual desde game_state.

start_game()
```
